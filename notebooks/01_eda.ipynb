```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_DIR = Path("../data/raw")

resume_path = DATA_DIR / "UpdatedResumeDataSet.csv"
resume_df = pd.read_csv(resume_path)

print("Dataset shape:", resume_df.shape)
display(resume_df.head())
print("Columns:", resume_df.columns.tolist())
```

```python
print("Missing values:")
display(resume_df.isnull().sum())

print("Duplicate rows:", resume_df.duplicated().sum())
print("Resume categories:", resume_df["Category"].nunique())

display(resume_df["Category"].value_counts().head(15))
```

```python
category_counts = resume_df["Category"].value_counts()

plt.figure(figsize=(12, 6))
category_counts.plot(kind="bar")
plt.title("Number of Resumes per Category")
plt.xlabel("Category")
plt.ylabel("Number of Resumes")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

```python
job_path = DATA_DIR / "naukri_com-job_sample.csv"
jobs_df = pd.read_csv(job_path)

print("Job dataset shape:", jobs_df.shape)
print("Job columns:", jobs_df.columns.tolist())
display(jobs_df.head())
display(jobs_df.isnull().sum().sort_values(ascending=False).head(10))
```
